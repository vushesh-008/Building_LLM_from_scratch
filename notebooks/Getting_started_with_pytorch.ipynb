```python
column_sum , row_sum = input.sum(dim= 0) , input.sum(dim=1) 
print("shape of column_sum: ", column_sum.shape) # Shape of the sum along rows (columns)
print("Sum along rows (dim=0): ", column_sum) # Sum along the first dimension (rows) - This will give us the sum of each column
print("shape of row_sum: ", row_sum.shape) # Shape of the sum along columns (rows)
print("Sum along columns (dim=1): ", row_sum) # Sum along the second dimension (columns) - This will give us the sum of each row
```

```python
column_sum_keep_dim , row_sum_keep_dim = input.sum(dim= 0, keepdim=True) , input.sum(dim=1, keepdim=True) 
print("shape of column_sum_keep_dim: ", column_sum_keep_dim.shape) # Shape of the sum along rows (columns) with keepdim=True
print("Sum along rows (dim=0) with keepdim=True: ", column_sum_keep_dim) # Sum along the first dimension (rows) with keepdim=True - This will give us
# the sum of each column while keeping the original number of dimensions
print("shape of row_sum_keep_dim: ", row_sum_keep_dim.shape) # Shape of
# the sum along columns (rows) with keepdim=True
print("Sum" \
" along columns (dim=1) with keepdim=True: ", row_sum_keep_dim) # Sum along the second dimension (columns) with keepdim=True - This will give us
# the sum of each row while keeping the original number of dimensions
```

```python
torch.manual_seed(123) # Set random seed for reproducibility
embeddings = torch.nn.Embedding(num_embeddings=10, embedding_dim=5) # Create an embedding layer with 10 embeddings of dimension 5
input_indices = torch.tensor([0, 7, 8, 3]) # Example input indices
output_embeddings = embeddings(input_indices) # Get the corresponding embeddings for the input indices
print("Output embeddings for input indices [0, 7, 8, 3]: \n", output_embeddings)

print(embeddings.weight) # Shape of the input indices
```

```python
matrix = torch.triu(torch.ones(5,5), diagonal=1).bool() # Create an upper triangular matrix of ones with shape (5, 5
print("Upper triangular boolean matrix: \n", matrix)

# Create a tensor of shape (5, 5) filled with random values

inputs = torch.randn(5, 5) # Create a tensor of shape (5, 5) filled with random values
print("Input tensor: \n", inputs)

inputs_masked = inputs.masked_fill(matrix, float('-inf')) # Mask the upper triangular part of the input tensor using the boolean matrix
print("Input tensor after masking with upper triangular matrix: \n", inputs_masked)
```

```python
dropout = torch.nn.Dropout(p=0.5) # Create a dropout layer with a dropout probability of 0.5
dropout_output = dropout(inputs) # Apply dropout to the input tensor
print("Output after applying dropout: \n", dropout_output)
```
